```python
from pathlib import Path

Path(".streamlit").mkdir(exist_ok=True)
Path(".streamlit/config.toml").write_text(
    '[theme]\n'
    'base = "light"\n'
    'primaryColor = "#9B7EC8"\n'
    'backgroundColor = "#FBF8FF"\n'
    'secondaryBackgroundColor = "#F3ECFB"\n'
    'textColor = "#3D3A50"\n'
    'font = "sans serif"\n\n'
    '[server]\n'
    'maxUploadSize = 300\n'
    'headless = true\n',
    encoding="utf-8",
)
print("Saved:", Path(".streamlit/config.toml").exists())
```
